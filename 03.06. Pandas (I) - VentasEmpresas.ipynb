# Operaciones básicas con data frames

Crea un programa llamado VentasEmpresas.py que cree un data frame con los datos de la siguiente tabla y los muestre por pantalla (puedes emplear tabulate para ello)



```python
import pandas as pd

datos_array = [
  ["Enero", 20600, 17900],
  ["Abril", 22500, 18500],
  ["Julio", 15400, 17600],
  ["Octubre", 21100, 18200]
]
df_datos_array = pd.DataFrame(datos_array, columns=["Mes", "Ventas", "Gastos"])

print(df_datos_array)
```

```python
from tabulate import tabulate

print(tabulate(df_datos_array, headers="keys", tablefmt="pretty"))
```

```python
datos_dict = {}
datos_dict["meses"] = ["Enero", "Abril", "Julio", "Octubre"]
datos_dict["ventas"] = [20600, 22500, 15400, 21100]
datos_dict["gastos"] = [17900, 18500, 17600, 18200]
df_datos_dict = pd.DataFrame(datos_dict)

print(df_datos_dict)
```

```python
print(tabulate(datos_dict, headers="keys", tablefmt="pretty"))
```
