- Crea un programa llamado LimpiarCSV.py y, partiendo del fichero CSV que se te proporcionará:
  - Homogeiniza la columna "nombre" para que todos los nombres estén capitalizados.
  - Detecta y elimina las filas duplicadas.
  - Limpia espacios sobrantes en la columna "provincia".

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("archivo.csv", sep=',')
df
```

```python
# Homogeneizar nombrees
df['nombre'] = df['nombre'].str.title()
df
```

```python
# Detectar y eliminar filas duplicadas
print(f"Datos duplicados: {df.duplicated()}")
df = df.drop_duplicates()
df
```

```python
# Eliminar espacios de provincias
# warning: df.loc['provincia'] = df['provincia'].str.strip()
df.loc[:, 'provincia'] = df['provincia'].str.strip()
df
```
