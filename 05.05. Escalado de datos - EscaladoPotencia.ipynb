- Crea un programa llamado EscaladoPotencia.py que utilice el archivo CSV que se te proporcionará sobre parques eólicos en Castilla y León y crea una nueva columna llamada `potencia_unitaria_norm` que normalice los valores de la columna `potencia_unitaria` entre 0 y 1.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("parques_eolicos.csv", sep=';')
df.head()
```

```python
potencia_unitaria_min = df['potencia_unitaria'].min()
potencia_unitaria_max = df['potencia_unitaria'].max()

print(f"Potencia unitaria mímina: {potencia_unitaria_min}")
print(f"Potencia unitaria máxima: {potencia_unitaria_max}")
```

```python
df['potencia_unitaria_norm'] = (df['potencia_unitaria'] - potencia_unitaria_min) / \
 (potencia_unitaria_max - potencia_unitaria_min)

df
```
