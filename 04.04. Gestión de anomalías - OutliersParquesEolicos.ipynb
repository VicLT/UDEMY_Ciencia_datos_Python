- Crea un programa llamado OutliersParquesEolicos.py que utilice el archivo CSV que se te proporciona sobre parques eólicos en Castilla y León e identifique los parques que sean anómalos en cuanto a potencia total.
- Aplica la técnica del IQR para identificarlos, tomando como umbral 0.5 veces el IQR.

```python
import pandas as pd
import numpy as np
```

```python
df_parques = pd.read_csv("parques_eolicos.csv", sep=';')
df_parques
```

```python
potencias = df_parques['potencia']
Q1, Q3 = np.percentile(potencias, [25, 75])
IQR = Q3 - Q1
outliers = []

for i in range(0, len(df_parques)):
  if df_parques.loc[i, 'potencia'] < (Q1 - 0.5 * IQR) or \
  df_parques.loc[i, 'potencia'] > (Q3 + 0.5 * IQR):
    outliers.append(df_parques.loc[i, :])

print(f"Potencias anómalas:\n{outliers}\n")
print(f"Total de parques eólicos: {len(df_parques)}")
print(f"Total de outliers: {len(outliers)}")
```
