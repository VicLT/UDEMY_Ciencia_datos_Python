# Operaciones avanzadas con data frames

Crea un programa llamado ParquesEolicosProvincia.py que utilice el archivo CSV que se te proporciona sobre parques eólicos de la comunidad de Castilla y León.


```python
import pandas as pd

df = pd.read_csv("parques_eolicos.csv", sep=";")
df
```

Nos interesa quedarnos con los parques de la provincia de Zamora que tengas más de 10 aerogeneradores.

```python
condicion = (df['provincia'] == 'ZA') & (df['aerogeneradores'] > 10)
df_parques_zamora = df[condicion]
df_parques_zamora
```

Muestra el listado resultante ordenado por número de aerogeneradores.

```python
df_parques_zamora = df_parques_zamora.sort_values(by='aerogeneradores', ascending=True)
df_parques_zamora
```

```python
df_parques_zamora = df_parques_zamora.reset_index(drop=True)
df_parques_zamora
```
