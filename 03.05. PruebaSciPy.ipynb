# SciPy

Disponemos de dos secuencias de datos:
* Temperaturas de una población: 22, 24, 26, 23, 25, 27, 28, 29, 34 y 36.
* Consumo de electricidad de esa población: 210, 220, 215, 225, 230, 240, 245, 250, 260 y 270.



```python
import numpy as np

temperaturas = np.array([22, 24, 26, 23, 25, 27, 28, 29, 34, 36])
electricidad = np.array([210, 220, 215, 225, 230, 240, 245, 250, 260, 270])
```

Calcula en un programa PruebaSciPy.py:
* Asimetría del conjunto de temperaturas
* Correlación entre temperatura y consumo (de los dos modos vistos)

```python
from scipy.stats import skew

asimetria_temps = skew(temperaturas)
print(f"Asimetría de temperaturas: {asimetria_temps}")
```

```python
from scipy.stats import pearsonr

coef_pearson, p_valor = pearsonr(temperaturas, electricidad)
print(f"Coeficiente de correlación de Pearson: {coef_pearson}")
print(f"Valor de que la correlación sea debida al azar: {p_valor}")
```

```python
from scipy.stats import spearmanr

coef_spearman, p_valor = spearmanr(temperaturas, electricidad)
print(f"Coeficiente de correlación de Spearman: {coef_spearman}")
print(f"Valor de que la correlación sea debida al azar: {p_valor}")
```
