Crea un programa CalculosEstadisticos.py y, partiendo del CSV que se proporciona:

```python
import pandas as pd
import numpy as np
from scipy.stats import skew
```

```python
df = pd.read_csv("players.csv", sep=',')
df.head()
```

- Elimina las filas que tengan valores nulos (calcula cuántas son).

```python
# Contar totalidad de filas antes de eliminar NaNs
filas_iniciales = len(df)
print(f"Cantidad de filas antes de la criba: {filas_iniciales}")
```

```python
# Eliminar filas con al menos un valor nulo
df = df.dropna()
```

```python
# Contar totalidad de filas después de eliminar NaNs
filas_finales = len(df)
print(f"Cantidad de filas después de la criba: {filas_finales}")
```

```python
# Cantidad de filas eliminadas
filas_eliminadas = filas_iniciales - filas_finales
print(f"Filas eliminadas: {filas_eliminadas}")
```

- Convierte la altura en pies (feet) en centímetros (1 pie = 30.48 cm.) en una nueva columna `height_cm`.

```python
df['height_cm'] = df['height_feet'] * 30.48
df.head()
```

- Convierte el peso en libras (pounds) en Kg (1 libra = 0.45 Kg) en una nueva columna `weight_kg`.

```python
df['weight_kg'] = (df['weight_pounds'] * 0.45).round(2)
df.head()
```

- Elimina las columnas `id`, `height_feet`, `height_inches` y `weight_pounds`.

```python
df = df.drop('id', axis=1)
df = df.drop('height_feet', axis=1)
df = df.drop('height_inches', axis=1)
df = df.drop('weight_pounds', axis=1)
df
```

- Averigua quién es el jugador más alto y el de menos peso.

```python
#jugador_mas_alto = df['height_cm'].max()
jugador_mas_alto = df.loc[df['height_cm'].idxmax()]
print(f"Jugador más alto:\n{jugador_mas_alto}")
```

```python
#jugador_menor_peso = df['weight_kg'].min()
jugador_menor_peso = df.loc[df['weight_kg'].idxmin()]
print(f"Jugador menor peso:\n{jugador_menor_peso}")
```

- Calcula la media, la mediana y la asimetría de las alturas e indica de qué tipo es dicha asimetría.

```python
media_altura = df['height_cm'].mean()
mediana_altura = df['height_cm'].median()
asimetria_altura = skew(df['height_cm'])
print(f"Media: {media_altura:.3f}")
print(f"Mediana: {mediana_altura:.3f}")
print(f"Asimetría: {asimetria_altura:.3f}")
```
