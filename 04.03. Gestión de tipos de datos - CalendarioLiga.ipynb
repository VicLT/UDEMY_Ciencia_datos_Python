```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_liga = pd.read_csv("liga_22_23.csv", sep=',')
df_liga.head()
```

```python
df_liga.info()
```

```python
df_liga['Date'] = pd.to_datetime(df_liga['Date'], format='%d-%m-%Y')
df_liga.info()
```

```python
df_liga = df_liga.set_index('Date').sort_index()
df_liga.head()
```

```python
octubre_22 = df_liga.loc['2022-10']
print(f"Cantidad partidos (filas) en octubre: {octubre_22.shape[0]}")
```
