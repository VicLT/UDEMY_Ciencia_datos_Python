```python
import pandas as pd

df = pd.read_csv("casas_rurales.csv", sep=";")
df
```

Queremos quedarnos solo con las 4 primeras columnas (id, localidad, codigo_postal y nombre), transformando el id y el codigo_postal a enteros de 32 bits.

```python
#df_lite = df.loc[:, ["id", "localidad", "codigo_postal", "nombre"]]
df_lite = df.iloc[:, 0:4]
df_lite
```

```python
print(df_lite.dtypes)
```

```python
df_lite = df_lite.astype({"id": "int32", "codigo_postal": "int32"})
print(df_lite.dtypes)
```

Guarda el resultado en un archivo llamado casas_rurales_resumen.csv

```python
df_lite.to_csv("casas_rurales_resumen.csv", index=False, sep=';')
```
