- Crea un programa llamado ProvinciasOneHot.py que utilice el CSV que se te proporciona sobre parques eólicos en Castilla y León, y codifica los valores de la columna "provincia" usando "one hot encoding", y eliminando la columna original. Muestra luego las 5 primeras filas para ver el resultado.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("parques_eolicos.csv", sep=';')
df.head()
```

```python
df = pd.get_dummies(
    df,
    columns=['provincia'],
    prefix='prov',
    dtype=int) # Sin dtype -> True/False en lugar de 1/0
df.head()
```
