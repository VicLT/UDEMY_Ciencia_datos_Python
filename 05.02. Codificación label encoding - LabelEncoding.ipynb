# Ejercicio propuesto

- A partir del fichero CSV que se te proporciona, crea un programa llamado LabelEncoding.py que codifique las siguientes columnas:
  - Las columnas "cholesterol" y "gluc" miden el nivel de colesterol y glucosa de los pacientes. Sus valores son "Normal", "Above Normal" y "Well Above Normal", y queremos codificarlos como 0, 1 y 2 respectivamente.
  - Las columnas "smoke", "alco", "active" y "cardio" indican si el paciente fuma, bebe alcohol, tiene una vida activa o si tiene problemas cardíacos, respectivamente. Queremos codificarlos como 0 (No) o 1 (Sí).
  - Reemplaza directamente la columna original con los nuevos valores codificados, en cada caso.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("datos_salud.csv", sep=',')
df
```

```python
scores_dict = {'Normal':0, 'Above Normal': 1, 'Well Above Normal': 2}
df['cholesterol'] = df['cholesterol'].map(scores_dict)
df['gluc'] = df['gluc'].map(scores_dict)
df.head()
```

```python
yes_no_dict = {'No': 0, 'Yes': 1}
df["smoke"] = df["smoke"].map(yes_no_dict)
df["alco"] = df["alco"].map(yes_no_dict)
df["active"] = df["active"].map(yes_no_dict)
df["cardio"] = df["cardio"].map(yes_no_dict)
df.head()
```
