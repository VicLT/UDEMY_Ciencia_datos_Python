# Uso de series

Crea un programa llamado SeriesNotas.py que cree una serie con la notas de unos alumnos (identificados por su código NIA), y luego muestre por pantalla las notas de los alumnos aprobados, ordenadas de mayor a menor.

```python
import pandas as pd

datos = {2222: 4.5, 4444: 9, 1111: 1.25, 3333: 10, 5555: 7.75}
alumnos_notas = pd.Series(datos)
aprobados = alumnos_notas[alumnos_notas >= 5]
aprobados_ordenados = aprobados.sort_values(ascending=False)
aprobados_ordenados
```
